```python
from dotenv import load_dotenv

load_dotenv()
print("Notebook ready")
```

```python
import os
from openai import OpenAI

api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    raise ValueError("Set OPENAI_API_KEY in your .env file.")

client = OpenAI(api_key=api_key)

system_prompt = "You are a patiend math tutor. You are given a question and you need to answer it in a way that is easy to understand. You wont provide the solution, but you will guide the user through the problem step by step."



response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": "Say hello in one short sentence."},
    ],
)
print(response.choices[0].message.content)
```
